```python
# Simulador de Batalha de Cartas (TCG)
# Desafio Semana 2 - Trilha de Python

# Função de ataque exigida no PDF
# Ela recebe os dados do ataque, reduz o HP do defensor e retorna o novo HP
def atacar(atacante, ataque, defensor, hp):

    # Subtrai o dano do HP do defensor
    hp -= ataque   # Forma mais "bonita# de hp = hp - ataque, pega o valor antigo e subtrai o dano para salvar a nova pontuação

    # Impede que o HP fique negativo
    if hp < 0:
        hp = 0

    # Mensagem narrando o ataque
    print(f"{atacante} atacou {defensor} causando {ataque} de dano!")

    # Retorna o novo HP atualizado
    return hp


# Função para mostrar o placar após cada turno
def exibir_placar(nome1, hp1, nome2, hp2):
    print("PLACAR")
    print(f"{nome1}: {hp1} HP")
    print(f"{nome2}: {hp2} HP")
    print("------------------") # Para dar um espaçamento entre os placares


# Entrada de dados:
# Dados do primeiro monstro
nome1 = input("Nome do Monstro 1: ")
hp1 = int(input("HP do Monstro 1: "))
atk1 = int(input("Ataque do Monstro 1: "))

# Dados do segundo monstro
nome2 = input("Nome do Monstro 2: ")
hp2 = int(input("HP do Monstro 2: "))
atk2 = int(input("Ataque do Monstro 2: "))


# Validação dos dados:
# O desafio pede que HP e ataque não sejam menores ou iguais a zero
if hp1 <= 0 or atk1 <= 0 or hp2 <= 0 or atk2 <= 0:

    print("Valores inválidos!")

else:

    # Variável para controlar os turnos
    turno = 1

    # Loop principal da batalha:
    # O duelo continua enquanto os dois monstros tiverem HP maior que zero
    while hp1 > 0 and hp2 > 0:

        # Mostrará o número do turno atual
        print(f"Turno {turno}")

        # Monstro 1 ataca primeiro
        hp2 = atacar(nome1, atk1, nome2, hp2)

        # Exibir o placar atualizado
        exibir_placar(nome1, hp1, nome2, hp2)

        65# Se o HP do monstro 2 chegou a zero, não faz sentido continuar o turno
        if hp2 == 0:
            break

        # Monstro 2 faz seu contra-ataca
        hp1 = atacar(nome2, atk2, nome1, hp1)

        # Exibir novamente o placar
        exibir_placar(nome1, hp1, nome2, hp2)

        # Fechou o turno e passa para o proximo
        turno += 1


    # Condição de vitória
    print(" Fim da Batalha ")

    # Verificando quem sobreviveu
    if hp1 > 0:
        print(f"Vencedor: {nome1}")
    else:
        print(f"Vencedor: {nome2}")
```

# 1. Qual a diferença entre for e while? Por que usar while no duelo?
O for é usado quando se sabe quantas vezes o código vai repetir. Enquanto o while é usado quando não se sabe quantas repetições serão necessárias. No duelo, o while foi melhor porque a quantidade de turnos quee serão necessários até um monstro perder toda a vida é incerto.

# 2. Para que serve o return em uma função?
Ele retorna um valor calculado pela função. Sem o return, o novo HP do monstro não seria salvo dentro da função definida.

# 3. O que é um loop infinito? Como evitá-lo?
Isso acontece quando o while nunca para de executar. Para evitar isso, é preciso garantir que a condição do while possa ficar "false" em algum momento.
No jogo em questão, os HPs diminuem a cada ataque, então a batalha termina quando um deles chegam a "zero".
