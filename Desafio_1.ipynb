```python
print("Sistema de Auxílio ao Orçamento")

# Cotação
cotacao_euro = 6.10

# Dados
orcamento = float(input("Qual o seu orçamento(BRL)?:"))
destino = input("Qual o destino da sua viagem?")
passagem_brl = float(input("Qual o valor da passagem(BRL):"))
hospedagem_por_dia_brl = float(input("Valor da hospedagem (por dia em BRL):"))
dias = int(input("Qual a duração da viagem(dias)?:"))

# Conversão
hospedagem_total_euro = hospedagem_por_dia_brl * dias * cotacao_euro
custo_total_euro = hospedagem_por_dia_brl * cotacao_euro

# Cálculo do custo total
custo_total_brl = passagem_brl + (hospedagem_por_dia_brl)*dias

# Validação de Orçamento
if custo_total_euro <= orcamento:
  status_orcamento = "Orçamento Possivel"
else:
  status_orcamento = "Orçamento não possivel"

# Status final da viagem
viavel = (custo_total_brl <= orcamento) and (dias > 0)
if viavel == True:
  status_viagem = "Viavel"
else:
  status_viagem = "Impossivel"

# Cálculo de quanto sobrará ou faltará para atingir o objetivo
sobrar_faltar = orcamento - custo_total_brl

# Exibição dos Resultados
print(f"Destino: {destino}")
print(f"Valor total da hospedagem (BRL): R$ {hospedagem_por_dia_brl:.2f}")
print(f"Custo total da viagem (BRL): R$ {custo_total_brl:.2f}")
print(f"Status da validação de orçamento: {status_orcamento}")
print(f"Viabilidade: {status_viagem}")
print(f"Quanto sobrou ou faltou para atingir o objetivo: R$ {sobrar_faltar:.2f}")
```
